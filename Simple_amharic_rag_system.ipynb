```python
from langchain_huggingface import HuggingFaceEmbeddings
# 1. Initialize the target embedding architectures

# Initialize the native Amharic embedding model
rasyosef_embebder = HuggingFaceEmbeddings(model_name="rasyosef/roberta-amharic-text-embedding-base")
# Initialize the multilingual embedding model
baai_embder = HuggingFaceEmbeddings(model_name="BAAI/bge-m3")
intfloat_embder = HuggingFaceEmbeddings(model_name="intfloat/multilingual-e5-large")
```

```python
import numpy as np
from itertools import combinations

def cosine_sim(a, b):
    a, b = np.array(a), np.array(b)
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))

# 4 topics x 6 sentences = 24 sentences total
sentences = {
    "history_war": [
        "የአድዋ ጦርነት በ1888 ዓ.ም በኢትዮጵያውያን እና ጣሊያኖች መካከል ተካሄደ።",
        "ኢትዮጵያውያን በአፄ ምኒልክ መሪነት በአድዋ ጦርነት ድል አደረጉ።",
        "የጣሊያን ወራሪ ኃይል በኢትዮጵያ ጦር ተሸነፈ።",
        "የአድዋ ድል የአፍሪካ ቅኝ ግዛት ታሪክ ትልቅ ምዕራፍ ነው።",
        "የኢትዮጵያ ጦር በዘመናዊ የጦር መሳሪያ ራሱን አደራጅቶ ተዋግቷል።",
        "የካቲት 23 ቀን የአድዋ ድል መታሰቢያ ቀን ሆኖ ይከበራል።",
    ],
    "food_drink": [
        "ቡና የኢትዮጵያ ባህላዊ መጠጥ ነው።",
        "እንጀራ ከጤፍ የሚሰራ ባህላዊ ምግብ ነው።",
        "ዶሮ ወጥ በበዓላት ወቅት የሚዘጋጅ ተወዳጅ ምግብ ነው።",
        "የቡና ስነ ስርዓት በኢትዮጵያ ባህል ውስጥ ትልቅ ቦታ አለው።",
        "ሽሮ ከምስር ዱቄት የሚዘጋጅ ቀላል ምግብ ነው።",
        "ጤፍ በኢትዮጵያ ከፍተኛ የሚመረት እህል ነው።",
    ],
    "weather_nature": [
        "ዛሬ በአዲስ አበባ ዝናብ ይዘንባል ተብሎ ይጠበቃል።",
        "የኢትዮጵያ ከፍተኛ ቦታዎች በክረምት ወቅት ቀዝቃዛ ናቸው።",
        "የሰሜን ተራሮች ብሔራዊ ፓርክ ውብ የተፈጥሮ ገጽታ አለው።",
        "የበልግ ወቅት አጭር ዝናብ የሚያመጣ ወቅት ነው።",
        "የሀገሪቱ የአየር ንብረት እንደ ከፍታ ይለያያል።",
        "ደመና በሰማይ ላይ ተንሰራፍቶ ፀሐይን ሸፈነው።",
    ],
    "education_school": [
        "ተማሪዎች በየቀኑ ጠዋት ወደ ትምህርት ቤት ይሄዳሉ።",
        "የመንግስት ትምህርት ቤቶች ነጻ ትምህርት ይሰጣሉ።",
        "መምህራን ለተማሪዎች የቤት ስራ ይሰጣሉ።",
        "ዩኒቨርሲቲዎች የከፍተኛ ትምህርት እድል ይሰጣሉ።",
        "ተማሪዎች ፈተና ከመውሰዳቸው በፊት ያጠናሉ።",
        "የትምህርት ጥራት ማሻሻል የመንግስት ቅድሚያ ተግባር ነው።",
    ],
}

all_sentences = [(topic, s) for topic, sents in sentences.items() for s in sents]
print(f"Total sentences: {len(all_sentences)}")

models = {
    "rasyosef": rasyosef_embebder,
    "bge-m3": baai_embder,
    "e5-large (query prefix)": intfloat_embder,
}

def get_embedding(embedder, text, model_name):
    if "e5" in model_name:
        text = f"query: {text}"
    return np.array(embedder.embed_query(text))

results = {}

for model_name, embedder in models.items():
    print(f"\nEmbedding with {model_name}...")
    vectors = [get_embedding(embedder, s, model_name) for _, s in all_sentences]

    within_sims = []
    across_sims = []

    for i, j in combinations(range(len(all_sentences)), 2):
        topic_i, _ = all_sentences[i]
        topic_j, _ = all_sentences[j]
        sim = cosine_sim(vectors[i], vectors[j])
        if topic_i == topic_j:
            within_sims.append(sim)
        else:
            across_sims.append(sim)

    within_mean = np.mean(within_sims)
    across_mean = np.mean(across_sims)
    gap = within_mean - across_mean

    results[model_name] = {
        "within_mean": within_mean,
        "within_std": np.std(within_sims),
        "across_mean": across_mean,
        "across_std": np.std(across_sims),
        "gap": gap,
    }

    print(f"  Within-topic sim:  {within_mean:.4f} (± {np.std(within_sims):.4f})")
    print(f"  Across-topic sim:  {across_mean:.4f} (± {np.std(across_sims):.4f})")
    print(f"  Separation gap:    {gap:.4f}")

print("\n" + "="*50)
print("SUMMARY (sorted by separation gap, higher = better)")
print("="*50)
for name, r in sorted(results.items(), key=lambda x: -x[1]["gap"]):
    print(f"{name:30s} gap={r['gap']:.4f}  (within={r['within_mean']:.3f}, across={r['across_mean']:.3f})")
```

## Generation: use CohereLabs/tiny-aya-earth as the RAG LLM

[`CohereLabs/tiny-aya-earth`](https://huggingface.co/CohereLabs/tiny-aya-earth) is a 3.35B-parameter model from the Tiny Aya family, tuned to be strongest on **West Asian and African languages** (including Amharic) — which makes it a good fit as the generator for this RAG pipeline. It needs `transformers` + `accelerate`, and (on a Colab GPU runtime) will run in fp16/bf16 for speed.

```python
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM, pipeline

llm_model_id = "CohereLabs/tiny-aya-earth"

tokenizer = AutoTokenizer.from_pretrained(llm_model_id)
model = AutoModelForCausalLM.from_pretrained(
    llm_model_id,
    torch_dtype=torch.bfloat16 if torch.cuda.is_available() else torch.float32,
    device_map="auto",
)

generator = pipeline(
    "text-generation",
    model=model,
    tokenizer=tokenizer,
)
```

### Build the retrieval + generation (RAG) function

We pick a retriever from the vectorstores built above — `vectorstore_baai` (BAAI/bge-m3) is used by default since it is a strong general multilingual embedder, but you can swap in `vectorstore_intfloat` or `vectorstore_rasyosef` based on the similarity comparisons in the cells above.

```python
retriever = vectorstore_baai.as_retriever(search_kwargs={"k": 3})

def build_context(question, k=3):
    docs = retriever.invoke(question)
    context = "\n\n".join(doc.page_content for doc in docs)
    return context, docs

def rag_answer(question, max_new_tokens=300, temperature=0.3):
    context, docs = build_context(question)

    system_prompt = (
        "You are a helpful assistant that answers questions in Amharic. "
        "Use ONLY the provided context to answer the question. "
        "If the answer is not contained in the context, say so honestly in Amharic."
    )
    user_prompt = f"አውድ (Context):\n{context}\n\nጥያቄ (Question): {question}\n\nመልስ (Answer):"

    messages = [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt},
    ]

    outputs = generator(
        messages,
        max_new_tokens=max_new_tokens,
        do_sample=temperature > 0,
        temperature=max(temperature, 1e-5),
        top_p=0.95,
    )

    reply = outputs[0]["generated_text"][-1]["content"]
    return reply, docs
```

### Try it out

```python
question = "የኢትዮጵያ ሰንደቅ ዓላማ ቀለማት ምንድናቸው?"
answer, retrieved_docs = rag_answer(question)

print("Question:", question)
print("\nRetrieved context:")
for i, d in enumerate(retrieved_docs, 1):
    print(f"[{i}] {d.page_content[:150]}...")

print("\nAnswer:\n", answer)
```

## Compare: with RAG vs. without RAG, evaluated with RAGAS

We now generate answers **with** retrieved context (the `rag_answer` function above) and **without** any context (the model answering from parametric memory alone), then score both with [RAGAS](https://docs.ragas.io/). RAGAS uses an LLM as a judge plus an embedding model — here we reuse the already-loaded `tiny-aya-earth` model and the `baai_embder` embeddings so everything stays local.

**Note:** `reference` below holds the ground-truth answers RAGAS compares against. The ones filled in are placeholders based on general knowledge of the Battle of Adwa — replace them with answers you've verified against your own `adwa.txt` before trusting the scores.

```python
def no_rag_answer(question, max_new_tokens=300, temperature=0.3):
    """Ask the model directly, with no retrieved context at all."""
    messages = [
        {"role": "system", "content": "You are a helpful assistant that answers questions in Amharic."},
        {"role": "user", "content": question},
    ]
    outputs = generator(
        messages,
        max_new_tokens=max_new_tokens,
        do_sample=temperature > 0,
        temperature=max(temperature, 1e-5),
        top_p=0.95,
    )
    return outputs[0]["generated_text"][-1]["content"]
```

```python
# Small eval set: question -> verified reference (ground-truth) answer.
# TODO: replace these reference answers with ones you've checked against your adwa.txt.
eval_set = [
    {
        "question": "የኢትዮጵያ ሰንደቅ ዓላማ ቀለማት ምንድናቸው?",
        "reference": "የኢትዮጵያ ሰንደቅ ዓላማ ቀለማት አረንጓዴ፣ ቢጫ እና ቀይ ናቸው።",
    },
    {
        "question": "የዓድዋ ጦርነት መቼ ተካሄደ?",
        "reference": "የዓድዋ ጦርነት እ.ኤ.አ. በ1896 (እ.ኤ.አ. መጋቢት 1) ተካሄደ።",
    },
    {
        "question": "በዓድዋ ጦርነት ኢትዮጵያን ማን መራ?",
        "reference": "ኢትዮጵያን በዓድዋ ጦርነት የመሩት አፄ ምኒልክ ዳግማዊ ናቸው።",
    },
]
```

```python
# Generate both sets of answers
with_rag_rows, without_rag_rows = [], []

for item in eval_set:
    q, ref = item["question"], item["reference"]

    rag_ans, rag_docs = rag_answer(q)
    with_rag_rows.append({
        "user_input": q,
        "retrieved_contexts": [d.page_content for d in rag_docs],
        "response": rag_ans,
        "reference": ref,
    })

    plain_ans = no_rag_answer(q)
    without_rag_rows.append({
        "user_input": q,
        "retrieved_contexts": [],       # no retrieval happened
        "response": plain_ans,
        "reference": ref,
    })

    print(f"Q: {q}\n  [RAG]    {rag_ans}\n  [No RAG] {plain_ans}\n")
```
